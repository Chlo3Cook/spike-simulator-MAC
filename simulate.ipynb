```python
import sys
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

HERE = Path.cwd()
sys.path.insert(0, str(HERE / "src"))
from simulator import simulate
from spike_detection import detect_spikes

data = np.load(HERE / "data" / "spike_data.npz")
t_real, v_real = data["t"], data["raw_data"]

t_sim, v_sim, true_times = simulate(10)

fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=True, sharey=True)
axes[0].plot(t_real, v_real, linewidth=0.5)
axes[0].set_title("Real recording")
axes[1].plot(t_sim, v_sim, linewidth=0.5, color="tab:orange")
axes[1].set_title(f"Simulated recording (seed 42): {len(true_times)} true spikes")
axes[1].set_xlabel("Time (s)")
for ax in axes:
    ax.set_ylabel("Voltage (a.u.)")
plt.tight_layout()
plt.show()

print(f"True spikes placed: {len(true_times)}")
print(f"Spikes your 2B detector finds in the simulation: {len(detect_spikes(v_sim, 1000))}")
```
